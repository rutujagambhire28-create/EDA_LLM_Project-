```python
import pandas as pd

data = {
    "Name": ["Amit", "Sneha", "Rahul", "Priya", "Neha",
             "Rohit", "Pooja", "Akash", "Kiran", "Sahil"],
    "Age": [20, 21, 19, 22, 20, 21, 19, 23, 20, 22],
    "Study_Hours": [2, 5, 3, 6, 4, 2, 7, 5, 3, 6],
    "Sleep_Hours": [8, 7, 6, 8, 7, 5, 8, 6, 7, 8],
    "Marks": [45, 85, 55, 90, 70, 40, 95, 80, 60, 88]
}

df = pd.DataFrame(data)

print("Dataset Created Successfully!")
print(df)
```

```python

# Display first 5 rows
print("First 5 Rows:")
print(df.head())

# Dataset dimensions
print("\nDataset Shape:")
print(df.shape)

# Column names
print("\nColumn Names:")
print(df.columns)

# Data types
print("\nData Types:")
print(df.dtypes)

# Statistical summary
print("\nStatistical Summary:")
print(df.describe())
```

```python
print("First 5 Rows:")
print(df.head())

print("\nLast 5 Rows:")
print(df.tail())

print("\nDataset Shape:")
print(df.shape)

print("\nColumn Names:")
print(df.columns)

print("\nData Types:")
print(df.dtypes)

print("\nStatistical Summary:")
print(df.describe())

```

```python

# Check missing values in each column
print("Missing Values:")
print(df.isnull().sum())
```

```python

# Check duplicate rows
print("Duplicate Rows:")
print(df.duplicated().sum())
```

```python

print("Dataset Information:")
df.info()
```

```python

# Remove duplicate rows
df = df.drop_duplicates()

# Fill missing numerical values with median
for col in df.select_dtypes(include="number").columns:
    df[col] = df[col].fillna(df[col].median())

# Fill missing text values with mode
for col in df.select_dtypes(include="object").columns:
    if df[col].isnull().any() and not df[col].mode().empty:
        df[col] = df[col].fillna(df[col].mode()[0])

print("Data Cleaning Completed Successfully!")
print(df.isnull().sum())
print("Dataset Shape:", df.shape)
```

```python

# Mean (Average)
print("Mean Marks:", df["Marks"].mean())

# Median (Middle Value)
print("Median Marks:", df["Marks"].median())

# Mode (Most Frequent Value)
print("Mode Marks:", df["Marks"].mode().tolist())

# Standard Deviation
print("Standard Deviation:", df["Marks"].std())

# Minimum Marks
print("Minimum Marks:", df["Marks"].min())

# Maximum Marks
print("Maximum Marks:", df["Marks"].max())
```

```python

print("Statistical Analysis of All Numerical Columns:")
print(df.describe())
```

```python

correlation = df["Study_Hours"].corr(df["Marks"])

print("Study Hours and Marks Correlation:", correlation)

if correlation > 0:
    print("Positive Relationship")
elif correlation < 0:
    print("Negative Relationship")
else:
    print("No Linear Relationship")
```
